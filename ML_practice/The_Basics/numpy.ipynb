```python
#multiplying two matrixes
a=np.array([[1,2,3]])
b=np.array([[1,2],[3,4],[5,6]])
print(b)
b.shape
a.shape
c=np.matmul(a,b)
print(c)
```

```python

#axis parameter tells the numpy Which direction to CRUSH 
#axis=0 ->move down (cols)
#axis=1 ->crush sideways(rows)
#you can also reOrganize the Arrays!
before=np.array([[1,2,3,4],[5,6,7,8]]) # again reshaping happens only if your shape is beyond 1d !1
print(before)

after=before.reshape((4,2)) #while shaping you must take care of the available number of digits so that it could fit in the shape you give 
print(after)
```

```python

#i can also stack arrays using np.vstack
a=np.array([1,2,3])
b=np.array([4,5,6])
print(np.vstack([a,b,b,b]))
```
